# Notebook 06 — Longitudinal Progress Tracker
Re-run as new data arrives. Tracks recovery and improvement over time for one user or group.

```
DATA_DIR    = "."
OUT_DIR     = "outputs"
USER_LABEL  = "Patient A"
ROLLING     = 3     # sessions to smooth over

import sys, os; sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import load_board_tries, load_bd_sessions, load_piano_sessions, save_fig, HAND_COLORS
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from scipy import stats

df_tries          = load_board_tries(DATA_DIR)
bd_valid, bd_bad  = load_bd_sessions(DATA_DIR)
piano_sess, _     = load_piano_sessions(DATA_DIR)
piano_valid       = piano_sess[piano_sess["is_valid"]].copy()

print(f"Board drawing sessions (valid): {len(bd_valid)}  |  Piano sessions (valid): {len(piano_valid)}")
```

## 1 · Fig 06a — Board drawing accuracy over sessions

```
bd_valid = bd_valid.reset_index(drop=True)
bd_valid["session_num"] = range(1, len(bd_valid)+1)
bd_valid["rolling_score"] = bd_valid["sessionScore"].rolling(ROLLING, min_periods=1).mean()

fig, ax = plt.subplots(figsize=(9,4))
ax.plot(bd_valid["session_num"], bd_valid["sessionScore"],
        "o--", color="#2c4fa0", alpha=0.5, label="Raw score")
ax.plot(bd_valid["session_num"], bd_valid["rolling_score"],
        lw=2.5, color="#2c4fa0", label=f"Rolling avg (n={ROLLING})")
# trend line
if len(bd_valid) >= 2:
    slope, intercept, r, p, _ = stats.linregress(bd_valid["session_num"], bd_valid["sessionScore"])
    x_line = np.array([1, len(bd_valid)])
    ax.plot(x_line, intercept + slope*x_line, "--", color="#c84b2f", lw=1.5,
            label=f"Trend (r²={r**2:.2f}, p={p:.2f})")
ax.set_xlabel("Session number"); ax.set_ylabel("Session score")
ax.set_title(f"Board drawing score over time — {USER_LABEL}", fontweight="bold")
ax.legend(); plt.tight_layout()
save_fig(fig, "fig06a_bd_score_over_sessions.png", OUT_DIR); plt.show()
```

## 2 · Fig 06b — Piano score over sessions

```
piano_valid = piano_valid.reset_index(drop=True)
piano_valid["session_num"] = range(1, len(piano_valid)+1)
piano_valid["rolling"] = piano_valid["sessionScore"].rolling(ROLLING, min_periods=1).mean()

fig, ax = plt.subplots(figsize=(9,4))
ax.plot(piano_valid["session_num"], piano_valid["sessionScore"],
        "o--", color="#ba7517", alpha=0.5, label="Raw score")
ax.plot(piano_valid["session_num"], piano_valid["rolling"],
        lw=2.5, color="#ba7517", label=f"Rolling avg (n={ROLLING})")
if len(piano_valid) >= 2:
    slope, intercept, r, p, _ = stats.linregress(piano_valid["session_num"], piano_valid["sessionScore"])
    x_line = np.array([1, len(piano_valid)])
    ax.plot(x_line, intercept + slope*x_line, "--", color="#c84b2f", lw=1.5,
            label=f"Trend (r²={r**2:.2f}, p={p:.2f})")
ax.set_xlabel("Session number"); ax.set_ylabel("Session score")
ax.set_title(f"Piano score over time — {USER_LABEL}", fontweight="bold")
ax.legend(); plt.tight_layout()
save_fig(fig, "fig06b_piano_score_over_sessions.png", OUT_DIR); plt.show()
```

## 3 · Fig 06c — Response time over sessions

```
fig, ax = plt.subplots(figsize=(9,4))
ax.plot(piano_valid["session_num"], piano_valid["avg_rt"],
        "o-", color="#2a6e4f", lw=2)
ax.set_xlabel("Session number"); ax.set_ylabel("Avg response time (s)")
ax.set_title(f"Piano response time over time — {USER_LABEL}
(should decrease as user improves)", fontweight="bold")
plt.tight_layout()
save_fig(fig, "fig06c_response_time_over_sessions.png", OUT_DIR); plt.show()
```

## 4 · Fig 06d — Hand asymmetry gap over sessions

```
df_sorted = df_tries.sort_values("startedAt").reset_index(drop=True)
asym_rows = []
# compute asymmetry in windows of 10 tries
window_size = 10
for start in range(0, len(df_sorted), window_size):
    chunk = df_sorted.iloc[start:start+window_size]
    l = chunk[chunk["hand"]=="Left"]["completed"].mean()*100 if (chunk["hand"]=="Left").sum() > 0 else np.nan
    r = chunk[chunk["hand"]=="Right"]["completed"].mean()*100 if (chunk["hand"]=="Right").sum() > 0 else np.nan
    asym_rows.append({"window": start//window_size+1, "asym": l-r if not (np.isnan(l) or np.isnan(r)) else np.nan})
asym_df = pd.DataFrame(asym_rows).dropna()

fig, ax = plt.subplots(figsize=(9,4))
ax.plot(asym_df["window"], asym_df["asym"], "o-", color="#2c4fa0", lw=2)
ax.axhline(0, color="black", lw=1, ls="--", label="No asymmetry")
ax.fill_between(asym_df["window"], asym_df["asym"], 0,
                where=asym_df["asym"]>0, alpha=0.15, color="#2a6e4f", label="Left better")
ax.fill_between(asym_df["window"], asym_df["asym"], 0,
                where=asym_df["asym"]<0, alpha=0.15, color="#c84b2f", label="Right better")
ax.set_xlabel(f"Try window (each = {window_size} tries)")
ax.set_ylabel("Left − Right success rate (pp)")
ax.set_title(f"Hand asymmetry over time — {USER_LABEL}
(gap should narrow toward 0 as impaired hand recovers)", fontweight="bold")
ax.legend(); plt.tight_layout()
save_fig(fig, "fig06d_hand_gap_over_sessions.png", OUT_DIR); plt.show()
```

## 5 · Fig 06e — Weekly summary heatmap

```
# Build a session × metric matrix (works even with a single day of data)
records = []
for i, row in bd_valid.iterrows():
    records.append({"session": f"BD-{i+1}", "score": row["sessionScore"],
                    "type": "Board Drawing", "date": row["time"]})
for i, row in piano_valid.iterrows():
    records.append({"session": f"Piano-{i+1}", "score": row["sessionScore"],
                    "type": "Piano", "date": row["time"]})

if records:
    rec_df = pd.DataFrame(records).sort_values("date").reset_index(drop=True)
    # Normalize scores per type to 0-1
    for t in rec_df["type"].unique():
        mask = rec_df["type"]==t
        mn = rec_df.loc[mask,"score"].min(); mx = rec_df.loc[mask,"score"].max()
        rec_df.loc[mask,"norm_score"] = (rec_df.loc[mask,"score"]-mn)/(mx-mn+1e-9)

    pivot = rec_df.pivot_table(index="session", columns="type", values="norm_score")
    fig, ax = plt.subplots(figsize=(6, max(3, len(pivot)*0.5+1)))
    im = ax.imshow(pivot.values, cmap="RdYlGn", aspect="auto", vmin=0, vmax=1)
    ax.set_xticks(range(len(pivot.columns))); ax.set_xticklabels(pivot.columns)
    ax.set_yticks(range(len(pivot.index)));   ax.set_yticklabels(pivot.index)
    plt.colorbar(im, ax=ax, label="Normalised score (per game type)")
    ax.set_title(f"Session performance heatmap — {USER_LABEL}", fontweight="bold")
    plt.tight_layout()
    save_fig(fig, "fig06e_session_heatmap.png", OUT_DIR); plt.show()
```

## 6 · Summary CSV

```
import warnings; warnings.filterwarnings("ignore")
summary_rows = []
for i, row in bd_valid.iterrows():
    summary_rows.append({"session_date": str(row.get("time",""))[:10],
                         "game":"board_drawing","score":row["sessionScore"]})
for i, row in piano_valid.iterrows():
    summary_rows.append({"session_date": str(row.get("time",""))[:10],
                         "game":"piano","score":row["sessionScore"],
                         "hit_rate": row["hit_rate"], "avg_rt": row["avg_rt"]})
pd.DataFrame(summary_rows).to_csv(f"{OUT_DIR}/longitudinal_summary.csv", index=False)
print("Saved longitudinal_summary.csv")
```
