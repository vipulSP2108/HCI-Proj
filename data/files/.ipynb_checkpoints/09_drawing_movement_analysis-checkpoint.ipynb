# Notebook 09 — Drawing Movement Analysis (Advanced)

Analyses the **kinematic properties** of drawing movements — not just whether shapes were completed,
but *how* they were drawn: smoothness, path efficiency, speed, and reaction time.

These metrics are directly relevant to clinical motor assessment and HCI research.

| Metric | What it measures | Clinical relevance |
|---|---|---|
| **Smoothness** (mean angle change, rad) | Consistency of direction — lower = smoother | Tremor, dysmetria, motor control quality |
| **Path efficiency** (target / drawn length) | How closely the drawn path matches target length | Impulsivity, undershoot/overshoot |
| **Drawing duration** (frames) | Time taken to complete the shape | Motor speed, fatigue |
| **Reaction time** (spawn → start drawing) | Cognitive readiness to begin | Cognitive-motor integration |
| **Point density** (n_points / duration) | How densely the path was sampled | Device consistency check |

| Audience | What they get |
|---|---|
| **User / Patient** | Are you drawing smoother and more efficiently over time? |
| **Doctor / Clinician** | Kinematic fingerprint: smoothness and efficiency as biomarkers |
| **HCI Researcher** | Shape-specific kinematics, learning curves, group comparisons |
| **App maker** | Which shapes produce the most extreme kinematics → tune difficulty |


```python
DATA_DIR    = '.'
OUT_DIR     = 'outputs'
GROUP_LABEL = 'Group A'
ROLLING_WIN = 5
import os, json, sys, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from scipy import stats as sp
warnings.filterwarnings('ignore')
os.makedirs(OUT_DIR, exist_ok=True)

import os
os.chdir(os.path.dirname(os.path.abspath(__file__))) if '__file__' in dir() else None

from hci_utils import (load_board_tries, load_bd_sessions, load_piano_sessions,
                        load_piano_movements, load_finger_movements,
                        build_key_position_map, compare_groups, save_fig,
                        SHAPE_ORDER, HAND_COLORS, GROUP_COLORS)

df = load_board_tries(DATA_DIR)
print(f"Loaded {len(df)} tries")
print(f"Columns: {list(df.columns)}")
print()
# Quick stats on new kinematic columns
cols = ['accuracy','path_length','target_length','path_efficiency','smoothness_rad','duration_frames','n_points']
print(df[cols].describe().round(3).to_string())
```

## 1 · Fig 09a — Smoothness per shape
Lower = smoother drawing. Failed tries tend to have higher (rougher) smoothness values.

```python
shapes = [s for s in SHAPE_ORDER if s in df['shapeType'].values]
fig, ax = plt.subplots(figsize=(10, 4))
data_comp = [df[(df['shapeType']==s) & (df['completed']==True)]['smoothness_rad'].dropna().values
             for s in shapes]
data_fail = [df[(df['shapeType']==s) & (df['completed']==False)]['smoothness_rad'].dropna().values
             for s in shapes]
x = np.arange(len(shapes)); w = 0.35

means_c = [d.mean() if len(d) else 0 for d in data_comp]
means_f = [d.mean() if len(d) else 0 for d in data_fail]
ax.bar(x - w/2, means_c, w, label='Completed', color='#2a6e4f', alpha=0.85)
ax.bar(x + w/2, means_f, w, label='Not completed', color='#c84b2f', alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(shapes, rotation=20, ha='right')
ax.set_ylabel('Mean smoothness (rad)\n(lower = smoother)')
ax.set_title(GROUP_LABEL + ' — Drawing smoothness per shape (completed vs failed)',
             fontweight='bold')
ax.legend()
plt.tight_layout()
save_fig(fig, 'fig09a_smoothness_per_shape.png', OUT_DIR)
plt.show()

print('\n=== CLINICAL INSIGHT ===')
roughest = shapes[int(np.argmax(means_c))] if any(means_c) else 'N/A'
smoothest = shapes[int(np.argmin([m if m > 0 else 99 for m in means_c]))] if any(means_c) else 'N/A'
print(f'Roughest completed shape: {roughest} — may indicate motor difficulty with this trajectory')
print(f'Smoothest completed shape: {smoothest} — user has good motor control for this shape')
print('\n=== RESEARCH NOTE ===')
print('Smoothness is a validated kinematic biomarker (Hogan & Sternad 2009).')
print('Mean jerk or spectral arc length are more rigorous — consider adding with more data.')
```

## 2 · Fig 09b — Path efficiency per shape
Efficiency = target path length / drawn path length.
- `~1.0` = ideal
- `< 1` = overdrawn (extra loops/strokes)
- `> 1` = underdrawn (gave up early)

```python
fig, ax = plt.subplots(figsize=(10, 4))
bp_data = [df[df['shapeType']==s]['path_efficiency'].replace([np.inf,-np.inf], np.nan).dropna().values
           for s in shapes]
bp = ax.boxplot(bp_data, labels=shapes, patch_artist=True,
                medianprops=dict(color='black', lw=2))
cmap = plt.cm.RdYlGn
for patch, vals in zip(bp['boxes'], bp_data):
    med = float(np.median(vals)) if len(vals) else 0
    # colour: green near 1.0, red when far
    dist = abs(med - 1.0)
    patch.set_facecolor(cmap(max(0, 1 - dist * 2)))
    patch.set_alpha(0.75)
ax.axhline(1.0, color='black', lw=1.5, ls='--', label='Ideal (efficiency=1.0)')
ax.axhline(0.5, color='#c84b2f', lw=1, ls=':', alpha=0.6, label='50% overdrawn')
ax.set_ylabel('Path efficiency (target len / drawn len)')
ax.set_title(GROUP_LABEL + ' — Path efficiency per shape', fontweight='bold')
ax.legend(fontsize=8)
ax.set_ylim(0, min(ax.get_ylim()[1], 6))
plt.tight_layout()
save_fig(fig, 'fig09b_path_efficiency.png', OUT_DIR)
plt.show()

print('\n=== CLINICAL INSIGHT ===')
for s, vals in zip(shapes, bp_data):
    if len(vals) == 0: continue
    med = float(np.median(vals))
    if med < 0.7:
        print(f'{s}: overdrawn (median eff={med:.2f}) — user draws more than needed, '
              'may indicate uncontrolled movement or tremor')
    elif med > 1.5:
        print(f'{s}: underdrawn (median eff={med:.2f}) — user gives up early on this shape')
```

## 3 · Fig 09c — Smoothness vs accuracy scatter
Each point is one drawing try. Reveals whether smoother drawing = better accuracy.

```python
valid = df[df['smoothness_rad'].notna() & df['accuracy'].notna()].copy()
fig, ax = plt.subplots(figsize=(8, 5))
colors = [HAND_COLORS.get(h, '#888') for h in valid['hand']]
sc = ax.scatter(valid['smoothness_rad'], valid['accuracy'],
                c=colors, alpha=0.65, s=60, edgecolors='white', lw=0.5)
ax.set_xlabel('Smoothness (rad) — lower is smoother')
ax.set_ylabel('Accuracy (%)')
ax.set_title(GROUP_LABEL + ' — Smoothness vs Accuracy\n(green=Left hand, red=Right hand)',
             fontweight='bold')
# correlation
if len(valid) >= 5:
    r, p = sp.pearsonr(valid['smoothness_rad'], valid['accuracy'])
    ax.text(0.05, 0.92,
            f'r={r:.2f}  p={p:.3f}  {"significant" if p<0.05 else "n.s."}',
            transform=ax.transAxes, fontsize=9, color='gray')
handles = [mpatches.Patch(color=v, label=k+' hand') for k,v in HAND_COLORS.items()]
ax.legend(handles=handles, fontsize=8)
plt.tight_layout()
save_fig(fig, 'fig09c_smoothness_vs_accuracy.png', OUT_DIR)
plt.show()

print('\n=== RESEARCH NOTE ===')
print('Negative correlation (smoother → higher accuracy) validates smoothness as a motor quality metric.')
print('Use this figure in your HCI paper to justify smoothness as a proxy for motor control quality.')
```

## 4 · Fig 09d — Drawing duration per shape and outcome

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Duration by shape
dur_data = [df[df['shapeType']==s]['duration_frames'].dropna().values for s in shapes]
axes[0].boxplot(dur_data, labels=shapes, patch_artist=True,
                medianprops=dict(color='black', lw=2),
                boxprops=dict(facecolor='#2c4fa0', alpha=0.5))
axes[0].set_ylabel('Duration (frames)')
axes[0].set_title('Drawing duration per shape', fontweight='bold')
axes[0].tick_params(axis='x', rotation=20)

# Duration: completed vs failed
dur_c = df[df['completed']==True]['duration_frames'].dropna()
dur_f = df[df['completed']==False]['duration_frames'].dropna()
bp2 = axes[1].boxplot([dur_c, dur_f], labels=['Completed','Failed'],
                       patch_artist=True, medianprops=dict(color='black', lw=2))
bp2['boxes'][0].set_facecolor('#2a6e4f'); bp2['boxes'][0].set_alpha(0.6)
bp2['boxes'][1].set_facecolor('#c84b2f'); bp2['boxes'][1].set_alpha(0.6)
axes[1].set_ylabel('Duration (frames)')
axes[1].set_title('Duration: Completed vs Failed', fontweight='bold')
if len(dur_c) >= 2 and len(dur_f) >= 2:
    _, p = sp.mannwhitneyu(dur_c, dur_f, alternative='two-sided')
    axes[1].text(0.5, 0.95, f'Mann-Whitney p={p:.3f}',
                 transform=axes[1].transAxes, ha='center', fontsize=9, color='gray')

plt.suptitle(GROUP_LABEL + ' — Drawing duration analysis', fontweight='bold')
plt.tight_layout()
save_fig(fig, 'fig09d_drawing_duration.png', OUT_DIR)
plt.show()

print('\n=== CLINICAL INSIGHT ===')
if len(dur_c) > 0 and len(dur_f) > 0:
    print(f'Completed tries: mean={dur_c.mean():.1f} frames  std={dur_c.std():.1f}')
    print(f'Failed tries:    mean={dur_f.mean():.1f} frames  std={dur_f.std():.1f}')
    if dur_f.mean() < dur_c.mean():
        print('Failed tries are shorter -> user gives up quickly, not just slow')
    else:
        print('Failed tries are longer -> user struggles but keeps trying')
```

## 5 · Fig 09e — Reaction time: spawn to start drawing
Extracted from the `play` event log in `boarddrawingsessions.json`.
Measures cognitive-motor readiness — how long after seeing the shape the user starts drawing.

```python
bd_valid, _ = load_bd_sessions(DATA_DIR)

if bd_valid['avg_reaction_frames'].isna().all():
    print('No reaction time data found in play event log.')
else:
    fig, axes = plt.subplots(1, 2, figsize=(11, 4))

    # distribution
    rt_vals = bd_valid['avg_reaction_frames'].dropna()
    axes[0].hist(rt_vals, bins=10, color='#2c4fa0', edgecolor='white', alpha=0.85)
    axes[0].axvline(rt_vals.mean(), color='#c84b2f', lw=2, ls='--',
                    label=f'Mean {rt_vals.mean():.1f} frames')
    axes[0].set_xlabel('Avg reaction time (frames)')
    axes[0].set_ylabel('Sessions')
    axes[0].set_title('Reaction time distribution', fontweight='bold')
    axes[0].legend()

    # over sessions
    bd_valid2 = bd_valid.reset_index(drop=True)
    bd_valid2['snum'] = range(1, len(bd_valid2)+1)
    axes[1].plot(bd_valid2['snum'], bd_valid2['avg_reaction_frames'],
                 'o-', color='#2c4fa0', lw=2)
    axes[1].set_xlabel('Session number')
    axes[1].set_ylabel('Avg reaction time (frames)')
    axes[1].set_title('Reaction time over sessions\n(decreasing = faster cognitive-motor readiness)',
                      fontweight='bold')

    plt.suptitle(GROUP_LABEL + ' — Cognitive-motor reaction time', fontweight='bold')
    plt.tight_layout()
    save_fig(fig, 'fig09e_reaction_time.png', OUT_DIR)
    plt.show()

    print('\n=== CLINICAL INSIGHT ===')
    print('Reaction time = time from shape appearing to pen-down.')
    print('Slow RT = cognitive processing delay or motor planning difficulty.')
    print('Improving RT over sessions = dual benefit: cognitive + motor recovery.')
    print('\n=== RESEARCH NOTE ===')
    print('This is a novel metric for tablet-based drawing tasks.')
    print('Compare RT between impaired and non-impaired populations for your HCI paper.')
```

## 6 · Fig 09f — Smoothness over time (learning curve)

```python
df_sorted = df.sort_values('createdAt').reset_index(drop=True)
df_sorted['try_num'] = range(1, len(df_sorted)+1)
df_sorted['rolling_smooth'] = df_sorted['smoothness_rad'].rolling(ROLLING_WIN, min_periods=1).mean()

fig, ax = plt.subplots(figsize=(10, 4))
for hand in ['Left','Right']:
    sub = df_sorted[df_sorted['hand']==hand].copy().reset_index(drop=True)
    if sub.empty: continue
    sub['idx'] = range(len(sub))
    sub['roll'] = sub['smoothness_rad'].rolling(ROLLING_WIN, min_periods=1).mean()
    ax.plot(sub['idx'], sub['roll'],
            color=HAND_COLORS.get(hand,'#888'), lw=2,
            label=f'{hand} hand (rolling n={ROLLING_WIN})')
    ax.scatter(sub['idx'], sub['smoothness_rad'],
               color=HAND_COLORS.get(hand,'#888'), alpha=0.25, s=20)

ax.set_xlabel('Try index (chronological)')
ax.set_ylabel('Smoothness (rad) — lower is better')
ax.set_title(GROUP_LABEL + ' — Smoothness learning curve over time', fontweight='bold')
ax.legend()
plt.tight_layout()
save_fig(fig, 'fig09f_smoothness_learning_curve.png', OUT_DIR)
plt.show()

print('\n=== RESEARCH NOTE ===')
print('Decreasing smoothness values over time = motor learning / recovery in progress.')
print('A plateau suggests the user has reached their current motor capacity.')
print('Sudden spike in smoothness = fatigue or symptom flare — worth flagging in clinical log.')
```

## 7 · Fig 09g — Path efficiency vs smoothness (kinematic fingerprint scatter)

```python
valid = df[df['path_efficiency'].notna() & df['smoothness_rad'].notna()].copy()
valid = valid[valid['path_efficiency'] < 10]  # exclude extreme outliers (abandoned tries)

fig, ax = plt.subplots(figsize=(8, 6))
for shape in shapes:
    sub = valid[valid['shapeType']==shape]
    if sub.empty: continue
    ax.scatter(sub['smoothness_rad'], sub['path_efficiency'],
               label=shape, alpha=0.7, s=60, edgecolors='white', lw=0.5)

ax.axhline(1.0, color='gray', lw=1, ls='--', alpha=0.5)
ax.axvline(valid['smoothness_rad'].median(), color='gray', lw=1, ls=':', alpha=0.5)
ax.set_xlabel('Smoothness (rad) — lower is smoother')
ax.set_ylabel('Path efficiency (target/drawn)')
ax.set_title(GROUP_LABEL + ' — Kinematic fingerprint: efficiency vs smoothness\n'
             'Top-left = ideal (smooth + efficient), bottom-right = rough + overdrawn',
             fontweight='bold')
ax.legend(fontsize=8, ncol=2)
plt.tight_layout()
save_fig(fig, 'fig09g_kinematic_fingerprint.png', OUT_DIR)
plt.show()

# Save full metrics CSV
metrics_cols = ['shapeType','hand','completed','accuracy','path_length','target_length',
                'path_efficiency','smoothness_rad','duration_frames','n_points']
df[metrics_cols].to_csv(os.path.join(OUT_DIR, 'drawing_kinematics.csv'), index=False)
print('drawing_kinematics.csv saved')
print('\n=== RESEARCH NOTE ===')
print('This scatter is the "kinematic fingerprint" of this user/session.')
print('Shapes in top-left = strong motor control. Shapes in bottom-right = impaired control.')
print('Overlay two groups on this plot (fig from notebook 05) for your main comparison figure.')
```
