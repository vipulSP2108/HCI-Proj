# Notebook 02 — Hand Asymmetry Analysis
Compares left vs right hand performance. The asymmetry score is the primary clinical recovery metric.

```
DATA_DIR    = "."
OUT_DIR     = "outputs"
GROUP_LABEL = "Group A"
ROLLING_WINDOW = 5

import sys, os; sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import load_board_tries, save_fig, SHAPE_ORDER, HAND_COLORS
import pandas as pd, numpy as np, matplotlib.pyplot as plt

df = load_board_tries(DATA_DIR)
print(f"Loaded {len(df)} tries")
df["hand"].value_counts()
```

## 1 · Per-hand summary

```
hand_summary = (
    df.groupby("hand")
    .agg(total_tries=("_id","count"), success_count=("completed","sum"),
         avg_accuracy=("accuracy","mean"), std_accuracy=("accuracy","std"))
    .reset_index()
)
hand_summary["success_rate"] = hand_summary["success_count"] / hand_summary["total_tries"] * 100
hand_summary["std_accuracy"]  = hand_summary["std_accuracy"].fillna(0)
print(hand_summary.to_string(index=False))
```

## 2 · Per-hand × shape matrix

```
hxs = (
    df.groupby(["hand","shapeType"])
    .agg(tries=("_id","count"), successes=("completed","sum"),
         avg_acc=("accuracy","mean"))
    .reset_index()
)
hxs["success_rate"] = hxs["successes"] / hxs["tries"] * 100
hxs.to_csv(f"{OUT_DIR}/hand_asymmetry_summary.csv", index=False)
print(hxs.to_string(index=False))
```

## 3 · Fig 02a — Grouped bar: success rate per shape per hand

```
shapes_present = [s for s in SHAPE_ORDER if s in df["shapeType"].values]
hands = df["hand"].unique()
x = np.arange(len(shapes_present)); width = 0.35

fig, ax = plt.subplots(figsize=(11, 5))
for i, hand in enumerate(sorted(hands)):
    vals = []
    for s in shapes_present:
        row = hxs[(hxs["hand"]==hand)&(hxs["shapeType"]==s)]
        vals.append(row["success_rate"].values[0] if len(row) else 0)
    ax.bar(x + i*width - width/2, vals, width, label=f"{hand} hand",
           color=HAND_COLORS.get(hand, "#888"), alpha=0.85, edgecolor="white")
ax.set_xticks(x); ax.set_xticklabels(shapes_present, rotation=20, ha="right")
ax.set_ylabel("Success rate (%)"); ax.set_ylim(0, 115)
ax.set_title(f"Success rate: Left vs Right hand per shape — {GROUP_LABEL}", fontweight="bold")
ax.legend(); ax.axhline(100, color="gray", ls=":", lw=1)
plt.tight_layout()
save_fig(fig, "fig02a_hand_success_rate.png", OUT_DIR); plt.show()
```

## 4 · Fig 02b — Box plot: accuracy distribution by hand

```
fig, ax = plt.subplots(figsize=(6, 4))
data   = [df[df["hand"]==h]["accuracy"].values for h in sorted(hands)]
labels = sorted(hands)
bp = ax.boxplot(data, labels=[f"{l} hand" for l in labels],
                patch_artist=True, medianprops=dict(color="black",lw=2))
for patch, lbl in zip(bp["boxes"], labels):
    patch.set_facecolor(HAND_COLORS.get(lbl, "#888")); patch.set_alpha(0.6)
ax.set_ylabel("Accuracy (%)")
ax.set_title(f"Accuracy distribution by hand — {GROUP_LABEL}", fontweight="bold")
ax.set_ylim(0, 110)
plt.tight_layout()
save_fig(fig, "fig02b_hand_accuracy_boxplot.png", OUT_DIR); plt.show()
```

## 5 · Fig 02c — Asymmetry score (diverging bar)

```
asym_rows = []
for s in shapes_present:
    sub = hxs[hxs["shapeType"]==s]
    l = sub[sub["hand"]=="Left"]["success_rate"].values
    r = sub[sub["hand"]=="Right"]["success_rate"].values
    score = (float(l[0]) - float(r[0])) if len(l) and len(r) else np.nan
    asym_rows.append({"shape": s, "asymmetry": score})
asym = pd.DataFrame(asym_rows).dropna()

fig, ax = plt.subplots(figsize=(8, 4))
colors = ["#2a6e4f" if v >= 0 else "#c84b2f" for v in asym["asymmetry"]]
ax.barh(asym["shape"], asym["asymmetry"], color=colors, height=0.5)
ax.axvline(0, color="black", lw=1)
ax.set_xlabel("Left success rate − Right success rate (%)
(positive = left is better)")
ax.set_title(f"Hand asymmetry score per shape — {GROUP_LABEL}", fontweight="bold")
plt.tight_layout()
save_fig(fig, "fig02c_asymmetry_score_per_shape.png", OUT_DIR); plt.show()
```

## 6 · Fig 02d — Accuracy over time per hand

```
df_s = df.sort_values("startedAt").reset_index(drop=True)
df_s["try_num"] = range(len(df_s))

fig, ax = plt.subplots(figsize=(10, 4))
for hand in sorted(hands):
    sub = df_s[df_s["hand"]==hand].copy().reset_index(drop=True)
    sub["rolling"] = sub["accuracy"].rolling(ROLLING_WINDOW, min_periods=1).mean()
    sub["idx"] = range(len(sub))
    ax.plot(sub["idx"], sub["rolling"], label=f"{hand} hand",
            color=HAND_COLORS.get(hand,"#888"), lw=2)
ax.set_xlabel(f"Try number (chronological, rolling window={ROLLING_WINDOW})")
ax.set_ylabel("Accuracy (%)")
ax.set_title(f"Accuracy over time by hand — {GROUP_LABEL}", fontweight="bold")
ax.legend(); ax.set_ylim(0,110)
plt.tight_layout()
save_fig(fig, "fig02d_hand_accuracy_over_time.png", OUT_DIR); plt.show()

# Overall asymmetry score
l_all = df[df["hand"]=="Left"]["success_rate"].mean() if "success_rate" in df.columns else df[df["hand"]=="Left"]["completed"].mean()*100
r_all = df[df["hand"]=="Right"]["completed"].mean()*100
l_all = df[df["hand"]=="Left"]["completed"].mean()*100
print(f"\nOverall asymmetry score: Left {l_all:.1f}% vs Right {r_all:.1f}% → gap = {l_all-r_all:.1f}pp")
```
