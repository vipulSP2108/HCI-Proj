# Notebook 07 — Session Quality & Engagement Flags
Checks abandonment rates, session length estimates, time-of-day patterns, and flags unusual sessions.

```
DATA_DIR    = "."
OUT_DIR     = "outputs"
GROUP_LABEL = "Group A"
MIN_KP      = 3    # piano sessions with fewer keypresses = abandoned
MIN_TRIES   = 1    # board sessions with score=0 = abandoned

import sys, os; sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import load_board_tries, load_bd_sessions, load_piano_sessions, save_fig
import pandas as pd, numpy as np, matplotlib.pyplot as plt

df_tries         = load_board_tries(DATA_DIR)
bd_valid, bd_bad = load_bd_sessions(DATA_DIR)
piano_sess, kp   = load_piano_sessions(DATA_DIR)

total_bd    = len(bd_valid) + len(bd_bad)
total_piano = len(piano_sess)
print(f"Board drawing — valid: {len(bd_valid)}, abandoned: {len(bd_bad)}, total: {total_bd}")
print(f"Piano — valid: {len(piano_sess[piano_sess['is_valid']])}, abandoned: {len(piano_sess[~piano_sess['is_valid']])}, total: {total_piano}")
```

## 1 · Fig 07a — Score distribution (all sessions)

```
all_scores = (
    list(bd_valid["sessionScore"]) +
    list(piano_sess[piano_sess["is_valid"]]["sessionScore"])
)
fig, ax = plt.subplots(figsize=(8,4))
ax.hist(all_scores, bins=20, color="#2c4fa0", edgecolor="white", alpha=0.85)
ax.set_xlabel("Session score"); ax.set_ylabel("Count")
ax.set_title(f"Session score distribution (all games) — {GROUP_LABEL}", fontweight="bold")
plt.tight_layout()
save_fig(fig, "fig07a_session_score_distribution.png", OUT_DIR); plt.show()
```

## 2 · Fig 07b — Abandonment rate

```
bd_abandoned   = len(bd_bad)
bd_valid_n     = len(bd_valid)
p_abandoned    = len(piano_sess[~piano_sess["is_valid"]])
p_valid_n      = len(piano_sess[piano_sess["is_valid"]])

fig, axes = plt.subplots(1,2,figsize=(8,4))
for ax, title, valid, abandoned in [
    (axes[0], "Board Drawing", bd_valid_n, bd_abandoned),
    (axes[1], "Piano",         p_valid_n,  p_abandoned),
]:
    if valid + abandoned == 0:
        ax.text(0.5,0.5,"No data",ha="center",va="center",transform=ax.transAxes)
    else:
        ax.pie([valid, abandoned],
               labels=[f"Valid ({valid})", f"Abandoned ({abandoned})"],
               colors=["#2a6e4f","#c84b2f"], autopct="%1.0f%%",
               startangle=90, wedgeprops=dict(edgecolor="white",lw=2))
    ax.set_title(title, fontweight="bold")
fig.suptitle(f"Session abandonment rate — {GROUP_LABEL}", fontweight="bold")
plt.tight_layout()
save_fig(fig, "fig07b_abandonment_rate.png", OUT_DIR); plt.show()
```

## 3 · Fig 07c — Time-of-day performance

```
bd_valid["hour"] = pd.to_datetime(bd_valid["time"], utc=True, errors="coerce").dt.hour
piano_sess["hour"] = pd.to_datetime(piano_sess["time"], utc=True, errors="coerce").dt.hour

fig, axes = plt.subplots(1,2,figsize=(11,4))
for ax, df_, score_col, title in [
    (axes[0], bd_valid,                          "sessionScore", "Board Drawing score by hour"),
    (axes[1], piano_sess[piano_sess["is_valid"]],"sessionScore", "Piano score by hour"),
]:
    if df_.empty or df_["hour"].isna().all():
        ax.text(0.5,0.5,"No data",ha="center",va="center",transform=ax.transAxes)
        continue
    hourly = df_.groupby("hour")[score_col].mean()
    ax.bar(hourly.index, hourly.values, color="#2c4fa0", alpha=0.8, width=0.7)
    ax.set_xlabel("Hour of day (UTC)"); ax.set_ylabel("Avg score")
    ax.set_title(title, fontweight="bold"); ax.set_xticks(range(0,24,2))
fig.suptitle(f"Time-of-day analysis — {GROUP_LABEL}", fontweight="bold")
plt.tight_layout()
save_fig(fig, "fig07c_time_of_day.png", OUT_DIR); plt.show()
```

## 4 · Fig 07d — Sessions per day

```
all_times = (
    list(bd_valid["time"]) +
    list(bd_bad["time"]) +
    list(piano_sess["time"])
)
all_times = pd.to_datetime(all_times, utc=True, errors="coerce").dropna()
days = all_times.dt.date.value_counts().sort_index()

fig, ax = plt.subplots(figsize=(9,3))
ax.bar(days.index.astype(str), days.values, color="#2c4fa0", alpha=0.85, width=0.5)
ax.set_xlabel("Date"); ax.set_ylabel("Number of sessions")
ax.set_title(f"Sessions per day — {GROUP_LABEL}", fontweight="bold")
plt.xticks(rotation=30, ha="right"); plt.tight_layout()
save_fig(fig, "fig07d_sessions_per_day.png", OUT_DIR); plt.show()
```

## 5 · Session quality flags CSV

```
flags = []
for _, row in bd_valid.iterrows():
    flags.append({"session_id": row.get("_id",""), "game":"board_drawing",
                  "score": row["sessionScore"], "is_abandoned": False,
                  "hour_of_day": row.get("hour", np.nan)})
for _, row in bd_bad.iterrows():
    flags.append({"session_id": row.get("_id",""), "game":"board_drawing",
                  "score": 0, "is_abandoned": True,
                  "hour_of_day": pd.to_datetime(row.get("time"), utc=True, errors="coerce").hour})
for _, row in piano_sess.iterrows():
    flags.append({"session_id": row["session_id"], "game":"piano",
                  "score": row["sessionScore"], "is_abandoned": not row["is_valid"],
                  "hour_of_day": row.get("hour", np.nan)})

flags_df = pd.DataFrame(flags)
flags_df.to_csv(f"{OUT_DIR}/session_quality_flags.csv", index=False)
print("Saved session_quality_flags.csv")
flags_df
```
