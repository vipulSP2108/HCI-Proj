# Notebook 04 — Piano Reaction Analysis
Analyses response time, hit/miss/timeout patterns, and key transition distances.

```
DATA_DIR    = "."
OUT_DIR     = "outputs"
GROUP_LABEL = "Group A"
MIN_KEYPRESSES = 3    # sessions with fewer are flagged as abandoned

import sys, os; sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import load_piano_sessions, load_piano_movements, save_fig
import pandas as pd, numpy as np, matplotlib.pyplot as plt

sessions, keypresses = load_piano_sessions(DATA_DIR)
movements = load_piano_movements(DATA_DIR)
valid_sessions = sessions[sessions["is_valid"]].copy()
print(f"Total sessions: {len(sessions)}  |  Valid: {len(valid_sessions)}  |  Abandoned: {len(sessions)-len(valid_sessions)}")
valid_sessions[["session_id","time","mode","sessionScore","hits","misses","timeouts","hit_rate","avg_rt"]]
```

## 1 · Fig 04a — Session scores over time

```
fig, ax = plt.subplots(figsize=(9,4))
colors = ["#2a6e4f" if r>=0.8 else "#ba7517" if r>=0.5 else "#c84b2f"
          for r in valid_sessions["hit_rate"].fillna(0)]
bars = ax.bar(range(len(valid_sessions)), valid_sessions["sessionScore"], color=colors, width=0.6)
ax.set_xticks(range(len(valid_sessions)))
ax.set_xticklabels([f"S{i+1}" for i in range(len(valid_sessions))])
ax.set_ylabel("Session score"); ax.set_title(f"Piano session scores — {GROUP_LABEL}", fontweight="bold")
from matplotlib.patches import Patch
ax.legend(handles=[Patch(color="#2a6e4f",label="Hit rate ≥80%"),
                   Patch(color="#ba7517",label="50–80%"),
                   Patch(color="#c84b2f",label="<50%")], fontsize=8)
plt.tight_layout()
save_fig(fig, "fig04a_piano_session_scores.png", OUT_DIR); plt.show()
```

## 2 · Fig 04b — Stacked hit/miss/timeout per session

```
fig, ax = plt.subplots(figsize=(9,4))
idx = range(len(valid_sessions))
ax.bar(idx, valid_sessions["hits"],     label="Hits",     color="#2a6e4f", width=0.6)
ax.bar(idx, valid_sessions["misses"],   label="Misses",   color="#c84b2f", width=0.6,
       bottom=valid_sessions["hits"])
ax.bar(idx, valid_sessions["timeouts"], label="Timeouts", color="#888780", width=0.6,
       bottom=valid_sessions["hits"]+valid_sessions["misses"])
ax.set_xticks(idx); ax.set_xticklabels([f"S{i+1}" for i in idx])
ax.set_ylabel("Keypresses"); ax.set_title(f"Hit / Miss / Timeout per session — {GROUP_LABEL}", fontweight="bold")
ax.legend(fontsize=9)
plt.tight_layout()
save_fig(fig, "fig04b_hit_miss_timeout.png", OUT_DIR); plt.show()
```

## 3 · Fig 04c — Response time histogram

```
valid_rt = keypresses[(keypresses["responsetime"]>0) & (keypresses["correct"]==1)]["responsetime"]
fig, ax = plt.subplots(figsize=(8,4))
ax.hist(valid_rt, bins=20, color="#2c4fa0", edgecolor="white", alpha=0.85)
ax.axvline(valid_rt.mean(),   color="#c84b2f", lw=2, ls="--", label=f"Mean {valid_rt.mean():.2f}s")
ax.axvline(valid_rt.median(), color="#2a6e4f", lw=2, ls="--", label=f"Median {valid_rt.median():.2f}s")
ax.set_xlabel("Response time (s)"); ax.set_ylabel("Count")
ax.set_title(f"Response time distribution (hits only) — {GROUP_LABEL}", fontweight="bold")
ax.legend(); plt.tight_layout()
save_fig(fig, "fig04c_response_time_histogram.png", OUT_DIR); plt.show()
```

## 4 · Fig 04d — Key transition distance vs accuracy

```
if movements.empty:
    print("No movement data found.")
else:
    bins   = [0, 200, 400, 600, 800, 1200]
    labels = ["0–200","200–400","400–600","600–800","800+"]
    movements["dist_bucket"] = pd.cut(movements["distance"], bins=bins, labels=labels)

    # merge correctness: match movements to session keypresses by session_id order
    # (movements and play arrays are parallel)
    mov_acc = movements.groupby("dist_bucket", observed=True).agg(count=("distance","count")).reset_index()

    # for a more precise join you'd need movement index — approximate here
    fig, ax = plt.subplots(figsize=(7,4))
    ax.bar(mov_acc["dist_bucket"].astype(str), mov_acc["count"], color="#2c4fa0", alpha=0.85, width=0.5)
    ax.set_xlabel("Key transition distance (px)"); ax.set_ylabel("Number of transitions")
    ax.set_title(f"Key transition distance distribution — {GROUP_LABEL}", fontweight="bold")
    plt.tight_layout()
    save_fig(fig, "fig04d_distance_distribution.png", OUT_DIR); plt.show()
```

## 5 · Fig 04e — Key transition heatmap (fromKey → toKey)

```
if not movements.empty:
    pivot = movements.groupby(["fromKey","toKey"]).size().unstack(fill_value=0)
    fig, ax = plt.subplots(figsize=(8,6))
    im = ax.imshow(pivot.values, cmap="Blues", aspect="auto")
    ax.set_xticks(range(len(pivot.columns))); ax.set_xticklabels(pivot.columns, rotation=45, ha="right")
    ax.set_yticks(range(len(pivot.index)));   ax.set_yticklabels(pivot.index)
    plt.colorbar(im, ax=ax, label="Transition count")
    ax.set_xlabel("To key"); ax.set_ylabel("From key")
    ax.set_title(f"Key transition frequency heatmap — {GROUP_LABEL}", fontweight="bold")
    plt.tight_layout()
    save_fig(fig, "fig04e_key_transition_heatmap.png", OUT_DIR); plt.show()
```
