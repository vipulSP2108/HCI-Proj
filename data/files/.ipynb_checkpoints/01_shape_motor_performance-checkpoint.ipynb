# Notebook 01 — Shape Motor Performance
Re-run this notebook whenever new data arrives. Point `DATA_DIR` at the folder containing the JSON files.

```
# ── CONFIG ─────────────────────────────────────────────
DATA_DIR  = "."          # folder with your JSON files
OUT_DIR   = "outputs"
GROUP_LABEL = "Group A"  # label shown in plot titles
ROLLING_WINDOW = 5       # tries to smooth over for trend lines
# ───────────────────────────────────────────────────────

import sys, os
sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import load_board_tries, save_fig, SHAPE_ORDER
import pandas as pd, numpy as np, matplotlib.pyplot as plt, matplotlib.patches as mpatches

df = load_board_tries(DATA_DIR)
print(f"Loaded {len(df)} tries | shapes: {df['shapeType'].unique()}")
df.head(3)
```

## 1 · Per-shape summary table

```
summary = (
    df.groupby("shapeType")
    .agg(
        total_tries   = ("_id",      "count"),
        success_count = ("completed","sum"),
        avg_accuracy  = ("accuracy", "mean"),
        std_accuracy  = ("accuracy", "std"),
    )
    .reset_index()
)
summary["success_rate"] = summary["success_count"] / summary["total_tries"] * 100
summary["std_accuracy"]  = summary["std_accuracy"].fillna(0)
# keep consistent order
cat = pd.CategoricalDtype(categories=SHAPE_ORDER, ordered=True)
summary["shapeType"] = summary["shapeType"].astype(cat)
summary = summary.sort_values("shapeType").reset_index(drop=True)
summary.to_csv(f"{OUT_DIR}/shape_summary.csv", index=False)
print("Saved shape_summary.csv")
summary
```

## 2 · Fig 01a — Success rate bar chart

```
fig, ax = plt.subplots(figsize=(9, 4))
colors = ["#2a6e4f" if v >= 80 else "#ba7517" if v >= 60 else "#c84b2f"
          for v in summary["success_rate"]]
bars = ax.barh(summary["shapeType"], summary["success_rate"], color=colors, height=0.55)
ax.axvline(80, color="#2a6e4f", lw=1, ls="--", alpha=0.5, label="80% threshold")
ax.axvline(60, color="#ba7517", lw=1, ls="--", alpha=0.5, label="60% threshold")
ax.set_xlabel("Success rate (%)")
ax.set_xlim(0, 110)
ax.set_title(f"Shape success rate — {GROUP_LABEL}", fontweight="bold")
for bar, val in zip(bars, summary["success_rate"]):
    ax.text(val + 1, bar.get_y() + bar.get_height()/2,
            f"{val:.1f}%", va="center", fontsize=9)
ax.legend(fontsize=8); plt.tight_layout()
save_fig(fig, "fig01a_shape_success_rate.png", OUT_DIR)
plt.show()
```

## 3 · Fig 01b — Accuracy distribution (box plot)

```
fig, ax = plt.subplots(figsize=(10, 4))
data = [df[df["shapeType"] == s]["accuracy"].values for s in SHAPE_ORDER
        if s in df["shapeType"].values]
labels = [s for s in SHAPE_ORDER if s in df["shapeType"].values]
bp = ax.boxplot(data, labels=labels, patch_artist=True,
                medianprops=dict(color="black", lw=2))
for patch, lbl in zip(bp["boxes"], labels):
    row = summary[summary["shapeType"] == lbl]
    v   = row["success_rate"].values[0] if len(row) else 50
    patch.set_facecolor("#2a6e4f" if v >= 80 else "#ba7517" if v >= 60 else "#c84b2f")
    patch.set_alpha(0.6)
ax.set_ylabel("Accuracy (%)")
ax.set_title(f"Accuracy distribution per shape — {GROUP_LABEL}", fontweight="bold")
ax.set_ylim(0, 110)
plt.tight_layout()
save_fig(fig, "fig01b_shape_accuracy_distribution.png", OUT_DIR)
plt.show()
```

## 4 · Fig 01c — Radar / spider chart

```
from matplotlib.patches import FancyArrowPatch
angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

sr_vals = []
ac_vals = []
for lbl in labels:
    row = summary[summary["shapeType"] == lbl]
    sr_vals.append(row["success_rate"].values[0] / 100 if len(row) else 0)
    ac_vals.append(row["avg_accuracy"].values[0] / 100 if len(row) else 0)
sr_vals += sr_vals[:1]; ac_vals += ac_vals[:1]

fig, ax = plt.subplots(figsize=(6, 6), subplot_kw=dict(polar=True))
ax.plot(angles, sr_vals, color="#2c4fa0", lw=2, label="Success rate")
ax.fill(angles, sr_vals, color="#2c4fa0", alpha=0.15)
ax.plot(angles, ac_vals, color="#c84b2f", lw=2, label="Avg accuracy")
ax.fill(angles, ac_vals, color="#c84b2f", alpha=0.15)
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=9)
ax.set_ylim(0, 1)
ax.set_yticks([0.25, 0.5, 0.75, 1.0])
ax.set_yticklabels(["25%","50%","75%","100%"], fontsize=7)
ax.set_title(f"Shape radar — {GROUP_LABEL}", fontweight="bold", pad=18)
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1), fontsize=9)
plt.tight_layout()
save_fig(fig, "fig01c_shape_radar.png", OUT_DIR)
plt.show()
```

## 5 · Fig 01d — Accuracy over time (rolling average per shape)

```
df_sorted = df.sort_values("startedAt").reset_index(drop=True)
df_sorted["try_num"] = df_sorted.groupby("shapeType").cumcount() + 1

fig, ax = plt.subplots(figsize=(10, 5))
cmap = plt.cm.get_cmap("tab10", len(labels))
for i, shape in enumerate(labels):
    sub = df_sorted[df_sorted["shapeType"] == shape].copy()
    if len(sub) < 2:
        continue
    sub["rolling_acc"] = sub["accuracy"].rolling(ROLLING_WINDOW, min_periods=1).mean()
    ax.plot(sub["try_num"], sub["rolling_acc"], label=shape,
            color=cmap(i), lw=2, marker="o", markersize=3)
ax.set_xlabel(f"Try number (per shape, rolling window={ROLLING_WINDOW})")
ax.set_ylabel("Accuracy (%)")
ax.set_title(f"Accuracy over time per shape — {GROUP_LABEL}", fontweight="bold")
ax.legend(fontsize=8, ncol=2); ax.set_ylim(0, 110)
plt.tight_layout()
save_fig(fig, "fig01d_shape_accuracy_over_time.png", OUT_DIR)
plt.show()
```

## 6 · Fig 01e — Try count vs success rate scatter

```
fig, ax = plt.subplots(figsize=(7, 4))
for _, row in summary.iterrows():
    ax.scatter(row["total_tries"], row["success_rate"], s=120,
               color="#2c4fa0", zorder=3)
    ax.annotate(row["shapeType"], (row["total_tries"], row["success_rate"]),
                textcoords="offset points", xytext=(6, 3), fontsize=8)
ax.set_xlabel("Number of tries")
ax.set_ylabel("Success rate (%)")
ax.set_title(f"Try count vs success rate — {GROUP_LABEL}", fontweight="bold")
ax.axhline(80, color="#2a6e4f", ls="--", lw=1, alpha=0.5)
ax.set_ylim(0, 110)
plt.tight_layout()
save_fig(fig, "fig01e_try_count_vs_success.png", OUT_DIR)
plt.show()
print("\n=== Summary ===")
print(summary[["shapeType","total_tries","success_rate","avg_accuracy","std_accuracy"]].to_string(index=False))
```
