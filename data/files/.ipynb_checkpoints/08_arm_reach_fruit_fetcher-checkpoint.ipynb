# Notebook 08 — Arm Reach / Fruit Fetcher (gamesessions.json)

**Status:** `gamesessions.json` is currently empty — the app is not yet writing session data for this game.

Every cell in this notebook handles the empty-file case gracefully and prints a clear message instead of crashing. Once the app starts logging, just re-run all cells.

### Expected schema (inferred from the app's other session files)
| Field | Description |
|---|---|
| `_id`, `user` | record and user IDs |
| `gameType`, `gameName` | e.g. `arm_reach` / `Arm Reach` |
| `time` | ISO timestamp |
| `sessionScore` | total score |
| `mode` | `mobile` or `laptop` |
| `play` | array of per-event objects with `correct`, `responsetime` |
| `coordinates` | arm movement path (x, y, timestamp) |
| `levelspan` | difficulty level |
| `systemMetrics` | `{userAgent, resolution}` |

Update the CONFIG cell below if your field names differ.

```
# ── CONFIG ─────────────────────────────────────────────
DATA_DIR         = '.'        # folder containing gamesessions.json
OUT_DIR          = 'outputs'
GROUP_LABEL      = 'Group A'

SCORE_FIELD      = 'sessionScore'
TIME_FIELD       = 'time'
USER_FIELD       = 'user'
MODE_FIELD       = 'mode'
PLAY_FIELD       = 'play'
COORD_FIELD      = 'coordinates'
LEVEL_FIELD      = 'levelspan'
CORRECT_FIELD    = 'correct'
RT_FIELD         = 'responsetime'
# ───────────────────────────────────────────────────────

import os, json, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
warnings.filterwarnings('ignore')
os.makedirs(OUT_DIR, exist_ok=True)

def save_fig(fig, fname):
    path = os.path.join(OUT_DIR, fname)
    fig.savefig(path, dpi=150, bbox_inches='tight')
    print(f'Saved -> {path}')

path = os.path.join(DATA_DIR, 'gamesessions.json')
raw  = json.load(open(path))

if not raw:
    print('gamesessions.json is currently empty.')
    print('Re-run this notebook once the app starts writing game session data.')
else:
    print(f'Loaded {len(raw)} records')
    print(f'Fields: {list(raw[0].keys())}')
```

## 1 · Build session DataFrame

```
if not raw:
    sessions = pd.DataFrame()
else:
    rows = []
    for s in raw:
        play     = s.get(PLAY_FIELD, [])
        hits     = sum(1 for p in play if p.get(CORRECT_FIELD) == 1)
        misses   = sum(1 for p in play if p.get(CORRECT_FIELD) == 0)
        timeouts = sum(1 for p in play if p.get(CORRECT_FIELD) == -1)
        valid_rt = [p[RT_FIELD] for p in play if p.get(RT_FIELD, 0) > 0]
        rows.append({
            'session_id'  : s.get('_id', ''),
            'user'        : s.get(USER_FIELD, ''),
            'time'        : pd.to_datetime(s.get(TIME_FIELD), utc=True, errors='coerce'),
            'mode'        : s.get(MODE_FIELD, 'unknown'),
            'gameType'    : s.get('gameType', 'unknown'),
            'gameName'    : s.get('gameName', 'unknown'),
            'score'       : s.get(SCORE_FIELD, 0),
            'level'       : s.get(LEVEL_FIELD, np.nan),
            'hits'        : hits,
            'misses'      : misses,
            'timeouts'    : timeouts,
            'total_events': len(play),
            'hit_rate'    : hits / len(play) if play else np.nan,
            'avg_rt'      : float(np.mean(valid_rt)) if valid_rt else np.nan,
            'is_valid'    : len(play) >= 3,
            'resolution'  : s.get('systemMetrics', {}).get('resolution', 'unknown'),
        })
    sessions = pd.DataFrame(rows).sort_values('time').reset_index(drop=True)
    sessions['session_num'] = range(1, len(sessions) + 1)
    print(f'Sessions: {len(sessions)}  |  Valid: {sessions["is_valid"].sum()}')
    display(sessions[['session_id','time','mode','gameType','score','hits','misses','hit_rate','avg_rt','is_valid']])
```

## 2 · Fig 08a — Session scores over time

```
if 'sessions' not in dir() or sessions.empty:
    print('No data yet.')
else:
    valid = sessions[sessions['is_valid']].copy()
    fig, ax = plt.subplots(figsize=(10, 4))
    colors = ['#2a6e4f' if r >= 0.8 else '#ba7517' if r >= 0.5 else '#c84b2f'
              for r in valid['hit_rate'].fillna(0)]
    ax.bar(valid['session_num'], valid['score'], color=colors, width=0.6)
    if len(valid) >= 2:
        from scipy import stats as sp
        sl, ic, rv, pv, _ = sp.linregress(valid['session_num'], valid['score'])
        xl = np.array([valid['session_num'].min(), valid['session_num'].max()])
        ax.plot(xl, ic + sl * xl, '--', color='black', lw=1.5, label=f'Trend (r2={rv**2:.2f})')
    ax.set_xlabel('Session'); ax.set_ylabel('Score')
    ax.set_title(f'Arm Reach / Fruit Fetcher — Session scores — {GROUP_LABEL}', fontweight='bold')
    ax.legend(handles=[
        mpatches.Patch(color='#2a6e4f', label='Hit rate >=80%'),
        mpatches.Patch(color='#ba7517', label='50-80%'),
        mpatches.Patch(color='#c84b2f', label='<50%'),
    ], fontsize=8)
    plt.tight_layout(); save_fig(fig, 'fig08a_gamesession_scores.png'); plt.show()
```

## 3 · Fig 08b — Hit / Miss / Timeout stacked bar

```
if 'sessions' not in dir() or sessions.empty:
    print('No data yet.')
else:
    valid = sessions[sessions['is_valid']].reset_index(drop=True)
    idx = range(len(valid))
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.bar(idx, valid['hits'],     label='Hits',     color='#2a6e4f', width=0.6)
    ax.bar(idx, valid['misses'],   label='Misses',   color='#c84b2f', width=0.6, bottom=valid['hits'])
    ax.bar(idx, valid['timeouts'], label='Timeouts', color='#888780', width=0.6,
           bottom=valid['hits'] + valid['misses'])
    ax.set_xticks(idx); ax.set_xticklabels([f'S{i+1}' for i in idx])
    ax.set_ylabel('Events'); ax.legend(fontsize=9)
    ax.set_title(f'Hit / Miss / Timeout per session — {GROUP_LABEL}', fontweight='bold')
    plt.tight_layout(); save_fig(fig, 'fig08b_hit_miss_timeout.png'); plt.show()
```

## 4 · Fig 08c — Response time distribution

```
if not raw:
    print('No data yet.')
else:
    all_rt = [p[RT_FIELD] for s in raw for p in s.get(PLAY_FIELD, [])
              if p.get(RT_FIELD, 0) > 0 and p.get(CORRECT_FIELD) == 1]
    if not all_rt:
        print('No response time data found in play records yet.')
    else:
        rt = pd.Series(all_rt)
        fig, ax = plt.subplots(figsize=(8, 4))
        ax.hist(rt, bins=20, color='#2c4fa0', edgecolor='white', alpha=0.85)
        ax.axvline(rt.mean(),   color='#c84b2f', lw=2, ls='--', label=f'Mean {rt.mean():.2f}s')
        ax.axvline(rt.median(), color='#2a6e4f', lw=2, ls='--', label=f'Median {rt.median():.2f}s')
        ax.set_xlabel('Reach time (s)'); ax.set_ylabel('Count')
        ax.set_title(f'Reach response time distribution — {GROUP_LABEL}', fontweight='bold')
        ax.legend(); plt.tight_layout()
        save_fig(fig, 'fig08c_response_time_histogram.png'); plt.show()
```

## 5 · Fig 08d — Arm movement path visualizer

Uses the `coordinates` field (x, y, timestamp per point).

```
if not raw:
    print('No data yet.')
else:
    has_coords = [s for s in raw if s.get(COORD_FIELD)]
    if not has_coords:
        print('No coordinate/path data found yet.')
        print('This figure will appear once the app writes arm movement paths.')
    else:
        CANVAS = 500
        fig, ax = plt.subplots(figsize=(6, 6))
        for s in has_coords[:20]:
            coords = s.get(COORD_FIELD, [])
            try:
                pts = np.array([[p['x'], p['y']] for p in coords])
                ax.plot(pts[:,0]*CANVAS, pts[:,1]*CANVAS, alpha=0.4, lw=1.5, color='#2c4fa0')
            except (KeyError, TypeError, IndexError):
                pass
        ax.set_xlim(0, CANVAS); ax.set_ylim(CANVAS, 0)
        ax.set_aspect('equal'); ax.axis('off')
        ax.set_title(f'Arm reach movement paths — {GROUP_LABEL}', fontweight='bold')
        plt.tight_layout(); save_fig(fig, 'fig08d_arm_reach_paths.png'); plt.show()
```

## 6 · Fig 08e — Performance by difficulty level

```
if 'sessions' not in dir() or sessions.empty or sessions['level'].isna().all():
    print('No level/difficulty data yet.')
else:
    valid = sessions[sessions['is_valid'] & sessions['level'].notna()]
    ls = valid.groupby('level').agg(
        avg_score=(   'score','mean'),
        avg_hit_rate=('hit_rate','mean'),
        count=('session_id','count')
    ).reset_index().sort_values('level')
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.bar(ls['level'].astype(str), ls['avg_score'], color='#2c4fa0', alpha=0.85, width=0.5)
    ax2 = ax.twinx()
    ax2.plot(range(len(ls)), ls['avg_hit_rate']*100, 'o--', color='#c84b2f', lw=2, label='Hit rate %')
    ax.set_xlabel('Difficulty level'); ax.set_ylabel('Avg score')
    ax2.set_ylabel('Avg hit rate (%)'); ax2.set_ylim(0, 110)
    ax.set_title(f'Performance by difficulty level — {GROUP_LABEL}', fontweight='bold')
    ax2.legend(fontsize=9); plt.tight_layout()
    save_fig(fig, 'fig08e_performance_by_level.png'); plt.show()
    print(ls.to_string(index=False))
```

## 7 · Fig 08f — Mobile vs Laptop comparison

```
if 'sessions' not in dir() or sessions.empty:
    print('No data yet.')
elif sessions['mode'].nunique() < 2:
    print(f'Only one mode so far: {sessions["mode"].unique()} — need both to compare.')
else:
    ms = sessions[sessions['is_valid']].groupby('mode').agg(
        avg_score=('score','mean'), avg_hit_rate=('hit_rate','mean'), avg_rt=('avg_rt','mean'),
        count=('session_id','count')
    ).reset_index()
    fig, axes = plt.subplots(1, 3, figsize=(11, 4))
    for ax, col, title in zip(axes,
        ['avg_score','avg_hit_rate','avg_rt'],
        ['Avg score','Avg hit rate','Avg response time (s)']):
        ax.bar(ms['mode'], ms[col], color=['#2c4fa0','#c84b2f'][:len(ms)], alpha=0.85, width=0.4)
        ax.set_title(title, fontweight='bold'); ax.set_xlabel('Mode')
    fig.suptitle(f'Mobile vs Laptop — {GROUP_LABEL}', fontweight='bold')
    plt.tight_layout(); save_fig(fig, 'fig08f_mode_comparison.png'); plt.show()
```

## 8 · Export summary CSV

```
if 'sessions' not in dir() or sessions.empty:
    cols = ['session_id','user','time','mode','gameType','gameName','score','level',
            'hits','misses','timeouts','total_events','hit_rate','avg_rt','is_valid','resolution']
    pd.DataFrame(columns=cols).to_csv(f'{OUT_DIR}/gamesession_summary_TEMPLATE.csv', index=False)
    print('Wrote gamesession_summary_TEMPLATE.csv  (shows expected columns when data arrives)')
else:
    sessions.to_csv(f'{OUT_DIR}/gamesession_summary.csv', index=False)
    print('Saved gamesession_summary.csv')
    print(sessions[['score','hits','misses','hit_rate','avg_rt']].describe().round(2))
```
