# Notebook 05 — Group Comparison
Compare two user groups (e.g. young vs elderly, patient vs control). Point each DATA_DIR at a folder of JSON files for that group.

```
# ── CONFIG ─────────────────────────────────────────────
DATA_DIR_A = "."         # folder for Group A
DATA_DIR_B = "."         # folder for Group B  ← change this
LABEL_A    = "Group A"
LABEL_B    = "Group B"
OUT_DIR    = "outputs"
# ───────────────────────────────────────────────────────

import sys, os; sys.path.insert(0, os.path.dirname(os.path.abspath("hci_utils.py")))
from hci_utils import (load_board_tries, load_piano_sessions,
                        compare_groups, save_fig, SHAPE_ORDER, GROUP_COLORS)
import pandas as pd, numpy as np, matplotlib.pyplot as plt

df_a = load_board_tries(DATA_DIR_A)
df_b = load_board_tries(DATA_DIR_B)
sess_a, _ = load_piano_sessions(DATA_DIR_A)
sess_b, _ = load_piano_sessions(DATA_DIR_B)

print(f"{LABEL_A}: {len(df_a)} tries  |  {LABEL_B}: {len(df_b)} tries")
```

## 1 · Fig 05a — Shape success rate comparison

```
shapes = [s for s in SHAPE_ORDER if s in df_a["shapeType"].values or s in df_b["shapeType"].values]

def shape_success(df):
    g = df.groupby("shapeType").agg(tries=("_id","count"), succ=("completed","sum")).reset_index()
    g["rate"] = g["succ"]/g["tries"]*100
    return dict(zip(g["shapeType"], g["rate"]))

sr_a = shape_success(df_a); sr_b = shape_success(df_b)
x = np.arange(len(shapes)); w=0.35

fig, ax = plt.subplots(figsize=(11,5))
ax.bar(x-w/2, [sr_a.get(s,0) for s in shapes], w, label=LABEL_A, color=GROUP_COLORS[0], alpha=0.85)
ax.bar(x+w/2, [sr_b.get(s,0) for s in shapes], w, label=LABEL_B, color=GROUP_COLORS[1], alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(shapes, rotation=20, ha="right")
ax.set_ylabel("Success rate (%)"); ax.set_ylim(0,115)
ax.set_title("Shape success rate: Group comparison", fontweight="bold")
ax.legend(); plt.tight_layout()
save_fig(fig, "fig05a_shape_success_comparison.png", OUT_DIR); plt.show()
```

## 2 · Fig 05b — Hand asymmetry comparison

```
def hand_asym(df):
    g = df.groupby("hand").agg(succ=("completed","sum"),tries=("_id","count")).reset_index()
    g["rate"] = g["succ"]/g["tries"]*100
    d = dict(zip(g["hand"],g["rate"]))
    return d.get("Left",np.nan) - d.get("Right",np.nan)

asym_a = hand_asym(df_a); asym_b = hand_asym(df_b)

fig, ax = plt.subplots(figsize=(5,4))
vals = [asym_a, asym_b]; labels = [LABEL_A, LABEL_B]
colors = [GROUP_COLORS[0], GROUP_COLORS[1]]
bars = ax.bar(labels, vals, color=colors, width=0.4, alpha=0.85)
ax.axhline(0, color="black", lw=1)
ax.set_ylabel("Left − Right success rate (pp)")
ax.set_title("Hand asymmetry score: Group comparison", fontweight="bold")
for bar, val in zip(bars, vals):
    ax.text(bar.get_x()+bar.get_width()/2, val+0.5, f"{val:.1f}pp", ha="center", fontsize=10)
plt.tight_layout()
save_fig(fig, "fig05b_hand_asymmetry_comparison.png", OUT_DIR); plt.show()
```

## 3 · Fig 05c — Accuracy distribution violin comparison

```
fig, ax = plt.subplots(figsize=(6,5))
parts_a = ax.violinplot([df_a["accuracy"].dropna()], positions=[1], showmedians=True)
parts_b = ax.violinplot([df_b["accuracy"].dropna()], positions=[2], showmedians=True)
for pc in parts_a["bodies"]: pc.set_facecolor(GROUP_COLORS[0]); pc.set_alpha(0.6)
for pc in parts_b["bodies"]: pc.set_facecolor(GROUP_COLORS[1]); pc.set_alpha(0.6)
ax.set_xticks([1,2]); ax.set_xticklabels([LABEL_A, LABEL_B])
ax.set_ylabel("Accuracy (%)"); ax.set_title("Accuracy distribution: Group comparison", fontweight="bold")
res = compare_groups(df_a["accuracy"], df_b["accuracy"], LABEL_A, LABEL_B)
ax.text(0.5,-0.12, f"p={res['p_value']:.3f}  effect size={res['effect_size']:.2f}  {'*significant*' if res['significant'] else 'not significant'}",
        transform=ax.transAxes, ha="center", fontsize=9, color="gray")
plt.tight_layout()
save_fig(fig, "fig05c_accuracy_distribution_comparison.png", OUT_DIR); plt.show()
```

## 4 · Fig 05d/e — Piano score and response time comparison

```
valid_a = sess_a[sess_a["is_valid"]]; valid_b = sess_b[sess_b["is_valid"]]

fig, axes = plt.subplots(1,2,figsize=(10,4))
for ax, col, title, fname in [
    (axes[0],"sessionScore","Piano session score","fig05d_piano_score_comparison.png"),
    (axes[1],"avg_rt","Avg response time (s)","fig05e_response_time_comparison.png"),
]:
    data = [valid_a[col].dropna(), valid_b[col].dropna()]
    bp = ax.boxplot(data, labels=[LABEL_A, LABEL_B], patch_artist=True,
                    medianprops=dict(color="black",lw=2))
    for patch, col_ in zip(bp["boxes"], GROUP_COLORS):
        patch.set_facecolor(col_); patch.set_alpha(0.6)
    ax.set_title(title, fontweight="bold")

plt.tight_layout()
save_fig(fig, "fig05d_piano_comparison.png", OUT_DIR); plt.show()
```

## 5 · Fig 05f — Radar summary comparison

```
def radar_vals(bd, piano):
    valid_p = piano[piano["is_valid"]]
    shapes_p = [s for s in SHAPE_ORDER if s in bd["shapeType"].values]
    avg_acc = bd["accuracy"].mean() / 100
    hand_gap = 1 - abs(hand_asym(bd)) / 100
    p_hit    = valid_p["hit_rate"].mean() if len(valid_p) else 0
    p_rt     = 1 - min(valid_p["avg_rt"].mean()/3, 1) if len(valid_p) else 0
    p_score  = min(valid_p["sessionScore"].mean()/120, 1) if len(valid_p) else 0
    return [avg_acc, hand_gap, p_hit, p_rt, p_score]

metrics = ["Avg accuracy","Hand symmetry","Piano hit rate","Response speed","Piano score"]
angles = np.linspace(0, 2*np.pi, len(metrics), endpoint=False).tolist(); angles += angles[:1]

va = radar_vals(df_a, sess_a); va += va[:1]
vb = radar_vals(df_b, sess_b); vb += vb[:1]

fig, ax = plt.subplots(figsize=(6,6), subplot_kw=dict(polar=True))
ax.plot(angles, va, color=GROUP_COLORS[0], lw=2, label=LABEL_A)
ax.fill(angles, va, color=GROUP_COLORS[0], alpha=0.15)
ax.plot(angles, vb, color=GROUP_COLORS[1], lw=2, label=LABEL_B)
ax.fill(angles, vb, color=GROUP_COLORS[1], alpha=0.15)
ax.set_xticks(angles[:-1]); ax.set_xticklabels(metrics, fontsize=9)
ax.set_ylim(0,1); ax.set_title("Overall performance radar", fontweight="bold", pad=18)
ax.legend(loc="upper right", bbox_to_anchor=(1.35,1.1))
plt.tight_layout()
save_fig(fig, "fig05f_radar_comparison.png", OUT_DIR); plt.show()
```

## 6 · Stats table

```
results = []
for col, da, db in [
    ("Accuracy (%)",     df_a["accuracy"],    df_b["accuracy"]),
    ("Piano score",      sess_a[sess_a["is_valid"]]["sessionScore"], sess_b[sess_b["is_valid"]]["sessionScore"]),
    ("Avg RT (s)",       sess_a[sess_a["is_valid"]]["avg_rt"],       sess_b[sess_b["is_valid"]]["avg_rt"]),
]:
    res = compare_groups(da, db, LABEL_A, LABEL_B)
    results.append({"Metric": col, **{k: round(v,3) if isinstance(v,float) else v for k,v in res.items()}})
pd.DataFrame(results).to_csv(f"{OUT_DIR}/group_comparison_stats.csv", index=False)
pd.DataFrame(results)
```
